```python
import matplotlib.pyplot as plt
import pandas as pd

smart_crop_data = {
    "yieldByCrop": [
        {"crop": "Rice", "yield": 4.8},
        {"crop": "Wheat", "yield": 4.1},
        {"crop": "Maize", "yield": 3.5},
        {"crop": "Cotton", "yield": 2.9}
    ],
    "rainfallVsYield": [
        {"week": "Week 1", "rainfall": 42, "yield": 4.6},
        {"week": "Week 2", "rainfall": 55, "yield": 4.9},
        {"week": "Week 3", "rainfall": 18, "yield": 3.9},
        {"week": "Week 4", "rainfall": 33, "yield": 4.2}
    ],
    "yieldTrend": [
        {"year": 2019, "yield": 3.7},
        {"year": 2020, "yield": 3.9},
        {"year": 2021, "yield": 4.1},
        {"year": 2022, "yield": 4.4},
        {"year": 2023, "yield": 4.6}
    ],
    "weeklyTemperature": [
        {"day": "Mon", "temp": 31},
        {"day": "Tue", "temp": 32},
        {"day": "Wed", "temp": 34},
        {"day": "Thu", "temp": 33},
        {"day": "Fri", "temp": 31},
        {"day": "Sat", "temp": 30},
        {"day": "Sun", "temp": 29}
    ]
}


df_yield = pd.DataFrame(smart_crop_data["yieldByCrop"])
df_rainfall = pd.DataFrame(smart_crop_data["rainfallVsYield"])
df_trend = pd.DataFrame(smart_crop_data["yieldTrend"])
df_temp = pd.DataFrame(smart_crop_data["weeklyTemperature"])

print(df_yield)
print(df_rainfall)
print(df_trend)
print(df_temp)


plt.show(block=True)
plt.figure(figsize=(6, 4))
plt.bar(df_yield["crop"], df_yield["yield"], color="#16a34a")
plt.title("Average Yield by Crop")
plt.ylabel("Tons per hectare")
plt.tight_layout()
plt.show()
```

```python
fig, ax1 = plt.subplots(figsize=(6, 4))
ax2 = ax1.twinx()

ax1.bar(df_rainfall["week"], df_rainfall["rainfall"], color="#0ea5e9", alpha=0.6, label="Rainfall (mm)")
ax2.plot(df_rainfall["week"], df_rainfall["yield"], color="#f97316", marker="o", label="Yield (t/ha)")

ax1.set_ylabel("Rainfall (mm)")
ax2.set_ylabel("Yield (t/ha)")
ax1.set_title("Rainfall vs Yield (Weekly)")
fig.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6, 4))
plt.plot(df_trend["year"], df_trend["yield"], marker="o", linewidth=2, color="#6366f1")
plt.title("Yearly Crop Yield Trend")
plt.xlabel("Year")
plt.ylabel("Yield (t/ha)")
plt.grid(True, linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6, 4))
plt.plot(df_temp["day"], df_temp["temp"], marker="o", linewidth=2, color="#ef4444")
plt.title("Weekly Temperature")
plt.xlabel("Day")
plt.ylabel("Temperature (°C)")
plt.grid(True, linestyle="--", alpha=0.4)
plt.tight_layout()  
plt.show()
```

```python
import matplotlib.pyplot as plt

crops = ["Rice", "Wheat", "Maize", "Cotton"]
yields = [4.8, 4.1, 3.5, 2.9]

plt.figure()
plt.bar(crops, yields)
plt.xlabel("Crop Type")
plt.ylabel("Yield (tons/ha)")
plt.title("Crop vs Yield")
plt.tight_layout()
plt.show()

```
